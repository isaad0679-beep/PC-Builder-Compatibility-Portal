```python
def display_components():

    # Load database
    with open("pc_builder_data/pc_components.json", "r") as file:
        data = json.load(file)

    print("\n========== PC COMPONENTS ==========")

    for category, items in data.items():

        print("\n", category.upper())
        print("-" * 40)

        for item in items:
            print("ID:", item["id"])
            print("Name:", item["name"])
            print("Price: Rs.", item["price"])
            print("-" * 30)


# Call the function
display_components()
```

```python
def search_record(search_name):

    # Load database
    with open("pc_builder_data/pc_components.json", "r") as file:
        data = json.load(file)

    found = False

    # Search in all categories
    for category, items in data.items():

        for item in items:

            if search_name.lower() in item["name"].lower():

                print("\nComponent Found!")
                print("Category:", category)
                print("ID:", item["id"])
                print("Name:", item["name"])
                print("Price: Rs.", item["price"])

                found = True

    if not found:
        print("Sorry! Component not found.")


# Take input from user
search_name = input("Enter component name to search: ")

# Call function
search_record(search_name)
```

```python
def add_record():

    try:
        # Load existing database
        with open("pc_builder_data/pc_components.json", "r") as file:
            data = json.load(file)

        print("\n========== ADD NEW COMPONENT ==========")
        print("\nAvailable Categories:")

        for category in data:
            print("-", category)

        # Take category input
        category_input = input("\nEnter component category: ").strip()

        # Match category without case sensitivity
        category = None

        for existing_category in data:
            if category_input.lower() == existing_category.lower():
                category = existing_category
                break

        if category is None:
            print("Invalid category! Please enter a valid category.")
            return

        # Take component details
        name = input("Enter component name: ").strip()

        if name == "":
            print("Component name cannot be empty!")
            return

        price = float(input("Enter component price: "))

        if price <= 0:
            print("Price must be greater than zero!")
            return

        # Generate unique ID
        new_id = max(
            item["id"]
            for items in data.values()
            for item in items
        ) + 1

        # Create new component
        new_component = {
            "id": new_id,
            "name": name,
            "price": price
        }

        # Take additional details
        if category in ["CPU", "Motherboard"]:
            socket = input("Enter socket (e.g. AM5): ").strip()
            new_component["socket"] = socket

        if category in ["Motherboard", "RAM"]:
            ram_type = input("Enter RAM type (DDR4/DDR5): ").strip().upper()
            new_component["ram_type"] = ram_type

        if category in ["CPU", "GPU", "Storage"]:
            wattage = int(input("Enter wattage: "))
            new_component["wattage"] = wattage

        # Add component to database
        data[category].append(new_component)

        # Save updated database
        with open("pc_builder_data/pc_components.json", "w") as file:
            json.dump(data, file, indent=4)

        print("\nComponent Added Successfully!")
        print("Component ID:", new_id)
        print("Component Name:", name)
        print("Category:", category)
        print("Price: Rs.", price)

    except ValueError:
        print("Invalid input! Please enter numbers correctly.")

    except FileNotFoundError:
        print("Database file not found!")

    except PermissionError:
        print("Permission denied! Cannot access database.")

    except Exception as e:
        print("An unexpected error occurred:", e)


# Run function
add_record()
```

```python
def check_compatibility():

    try:
        # Load database
        with open("pc_builder_data/pc_components.json", "r") as file:
            data = json.load(file)

        print("\n========== CPU COMPATIBILITY CHECKER ==========")

        # Display CPUs
        print("\nAvailable CPUs:")

        for cpu in data["CPU"]:
            print(cpu["id"], "-", cpu["name"], "| Socket:", cpu["socket"])

        # Select CPU
        cpu_id = int(input("\nEnter CPU ID: "))

        selected_cpu = None

        for cpu in data["CPU"]:
            if cpu["id"] == cpu_id:
                selected_cpu = cpu
                break

        if selected_cpu is None:
            print("CPU not found!")
            return

        # Display Motherboards
        print("\nAvailable Motherboards:")

        for motherboard in data["Motherboard"]:
            print(
                motherboard["id"], "-",
                motherboard["name"],
                "| Socket:", motherboard["socket"]
            )

        # Select motherboard
        motherboard_id = int(input("\nEnter Motherboard ID: "))

        selected_motherboard = None

        for motherboard in data["Motherboard"]:
            if motherboard["id"] == motherboard_id:
                selected_motherboard = motherboard
                break

        if selected_motherboard is None:
            print("Motherboard not found!")
            return

        # Check socket compatibility
        print("\n========== COMPATIBILITY RESULT ==========")

        if selected_cpu["socket"] == selected_motherboard["socket"]:

            print("CPU:", selected_cpu["name"])
            print("Motherboard:", selected_motherboard["name"])
            print("Result: COMPATIBLE")

        else:

            print("CPU:", selected_cpu["name"])
            print("Motherboard:", selected_motherboard["name"])
            print("Result: NOT COMPATIBLE")

    except ValueError:
        print("Please enter valid numeric IDs.")

    except FileNotFoundError:
        print("Database file not found!")

    except PermissionError:
        print("Permission denied!")

    except Exception as e:
        print("Unexpected error:", e)


# Run compatibility checker
check_compatibility()
```

```python
def check_ram_compatibility():

    try:
        # Load database
        with open("pc_builder_data/pc_components.json", "r") as file:
            data = json.load(file)

        print("\n========== RAM COMPATIBILITY CHECKER ==========")

        # Display RAM options
        print("\nAvailable RAM:")

        for ram in data["RAM"]:
            print(ram["id"], "-", ram["name"],
                  "| Type:", ram["ram_type"])

        # Select RAM
        ram_id = int(input("\nEnter RAM ID: "))

        selected_ram = None

        for ram in data["RAM"]:
            if ram["id"] == ram_id:
                selected_ram = ram
                break

        if selected_ram is None:
            print("RAM not found!")
            return

        # Display motherboards
        print("\nAvailable Motherboards:")

        for motherboard in data["Motherboard"]:
            print(motherboard["id"], "-", motherboard["name"],
                  "| RAM Type:", motherboard["ram_type"])

        # Select motherboard
        motherboard_id = int(input("\nEnter Motherboard ID: "))

        selected_motherboard = None

        for motherboard in data["Motherboard"]:
            if motherboard["id"] == motherboard_id:
                selected_motherboard = motherboard
                break

        if selected_motherboard is None:
            print("Motherboard not found!")
            return

        # Compare RAM types
        print("\n========== COMPATIBILITY RESULT ==========")

        if selected_ram["ram_type"] == selected_motherboard["ram_type"]:

            print("RAM:", selected_ram["name"])
            print("Motherboard:", selected_motherboard["name"])
            print("Result: COMPATIBLE")

        else:

            print("RAM:", selected_ram["name"])
            print("Motherboard:", selected_motherboard["name"])
            print("Result: NOT COMPATIBLE")

    except ValueError:
        print("Please enter valid numeric IDs.")

    except FileNotFoundError:
        print("Database file not found!")

    except PermissionError:
        print("Permission denied!")

    except Exception as e:
        print("Unexpected error:", e)


# Run RAM compatibility checker
check_ram_compatibility()
```

```python
def calculate_budget():

    try:
        # Load database
        with open("pc_builder_data/pc_components.json", "r") as file:
            data = json.load(file)

        print("\n========== PC BUILD BUDGET CALCULATOR ==========")

        selected_components = []
        total_price = 0

        # Select one component from each category
        for category, items in data.items():

            print("\nAvailable", category, ":")

            for item in items:
                print(
                    item["id"], "-",
                    item["name"],
                    "| Rs.", item["price"]
                )

            component_id = int(input("\nEnter " + category + " ID: "))

            selected_component = None

            for item in items:
                if item["id"] == component_id:
                    selected_component = item
                    break

            if selected_component is None:
                print("Invalid component ID!")
                return

            selected_components.append(selected_component)

            total_price += selected_component["price"]

        # Display final PC build
        print("\n========== YOUR PC BUILD ==========")

        for component in selected_components:
            print(
                component["name"],
                "- Rs.",
                component["price"]
            )

        print("-" * 40)
        print("TOTAL PC BUILD COST: Rs.", total_price)
        print("Budget Calculation Completed Successfully!")

    except ValueError:
        print("Invalid input! Please enter numeric IDs.")

    except FileNotFoundError:
        print("Database file not found!")

    except PermissionError:
        print("Permission denied!")

    except Exception as e:
        print("Unexpected error:", e)


# Run budget calculator
calculate_budget()
```

```python
def main_menu():

    while True:

        print("\n")
        print("=" * 50)
        print("       PC BUILDER & COMPATIBILITY PORTAL")
        print("=" * 50)

        print("1. Display All Components")
        print("2. Search Component")
        print("3. Add New Component")
        print("4. CPU Compatibility Checker")
        print("5. RAM Compatibility Checker")
        print("6. PC Build Budget Calculator")
        print("7. Generate PC Build Report")
        print("8. Exit Portal")

        print("=" * 50)

        try:
            choice = int(input("Enter your choice (1-8): "))

            if choice == 1:
                display_components()

            elif choice == 2:
                search_name = input("Enter component name: ")
                search_record(search_name)

            elif choice == 3:
                add_record()

            elif choice == 4:
                check_compatibility()

            elif choice == 5:
                check_ram_compatibility()

            elif choice == 6:
                calculate_budget()

            elif choice == 7:
                generate_report()

            elif choice == 8:
                print("\nThank you for using PC Builder Portal!")
                print("Portal Closed Successfully.")
                break

            else:
                print("Invalid choice! Please select 1-8.")

        except ValueError:
            print("Invalid input! Please enter a number.")
```

```python
def user_login():
    print("\n========== PC BUILDER LOGIN ==========")

    correct_username = "admin"
    correct_password = "1234"

    for attempt in range(3):
        username = input("Enter username: ")
        password = input("Enter password: ")

        if username == correct_username and password == correct_password:
            print("\nLogin Successful!")
            return True
        else:
            print("\nIncorrect username or password!")
            print("Attempts remaining:", 2 - attempt)

    print("\nToo many failed attempts. Access Denied.")
    return False


if user_login():
    main_menu()
```

# PC Builder & Compatibility Checker Portal

## Project Overview
The PC Builder & Compatibility Checker Portal is a Python-based system that helps users select computer components and check their compatibility.

1. User Login System
2. Display All Components
3. Search Components
4. Add New Components
5. CPU and Motherboard Compatibility Checker
6. RAM Compatibility Checker
7. PC Build Budget Calculator
8. Generate PC Build Report
